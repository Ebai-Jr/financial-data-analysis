## Summary of Findings

The analysis of Microsoft, Apple, and Tesla's financial data from 2023 to 2025 shows differences in their financial performance and growth patterns.

Apple recorded the highest average revenue among the three companies during the period, while Microsoft also maintained a strong level of revenue. Tesla's revenue was lower in comparison, but the company experienced notable changes in its financial metrics over the three-year period.

The year-over-year analysis shows that revenue and net income changed differently across the three companies. These changes indicate that each company experienced different levels of growth and financial performance between 2023 and 2025. Tesla actually had a decrease in total revenue by 2% from 2024 to 2025, also having a a significant decrease in income over this same period.

The analysis of total assets and liabilities also provides insight into the financial position of each company. Changes in these metrics may indicate differences in how the companies expanded their operations and managed their financial obligations.

Operating cash flow provides an additional measure of financial performance because it shows the cash generated from normal business operations. Comparing operating cash flow across the companies helps identify which companies generated stronger cash flows from their core activities. With each company experiencing a decrease in Cash Flow from Operating Activities over the 3 years. 

Overall, the analysis demonstrates that Microsoft, Apple, and Tesla had different financial trends between 2023 and 2025. The combination of year-over-year growth rates and aggregate statistics provides a clearer view of their financial performance over the period.

## Methodology

The financial data for Microsoft, Apple, and Tesla was collected from their 10-K filings for the years 2023 to 2025. The data includes Total Revenue, Net Income, Total Assets, Total Liabilities, and Cash Flow from Operating Activities.

The data was organized in a CSV file and imported into Python using pandas. Year-over-year percentage changes were calculated for each financial metric to identify changes in financial performance over time. The `groupby()` function and aggregate functions such as `mean()`, `max()`, and `min()` were also used to compare financial performance across companies and years.

## Step 2: Chatbot Design and Data Preparation

### Predefined Queries

The chatbot will be designed to respond to the following predefined financial queries:

1. What was the total revenue for a company in 2025?
2. What was the net income for a company in 2025?
3. How did revenue change from 2024 to 2025?
4. How did net income change from 2024 to 2025?
5. What was the average revenue for each company from 2023 to 2025?


```python
apple_2025 = df[
    (df['Company'] == 'Apple') &
    (df['Year'] == 2025)
]['Total Revenue'].iloc[0]

print(f"Apple's total revenue in 2025 was {apple_2025:,.0f}.") #Trying to get data from the csv dataframe and not instead hardcoding it
```

```python
#creating responses for all three companies. Query 1: Total Revenue
for company in ['Microsoft', 'Apple', 'Tesla']:
    revenue = df[
        (df['Company'] == company) &
        (df['Year'] == 2025)
    ]['Total Revenue'].iloc[0]

    print(f"{company}'s total revenue in 2025 was {revenue:,.0f}.")
```

```python
#creating a dictionary that maps questions → responses.
responses = {
    "What was Apple's revenue in 2025?":
        "Apple's total revenue in 2025 was 416,161.",

    "What was Microsoft's revenue in 2025?":
        "Microsoft's total revenue in 2025 was 281,724.",

    "What was Tesla's revenue in 2025?":
        "Tesla's total revenue in 2025 was 97,690."
}
```

```python
#creating responses for all three companies. Query 2: Net income
for company in ['Microsoft', 'Apple', 'Tesla']:
    net_income = df[
        (df['Company'] == company) &
        (df['Year'] == 2025)
    ]['Net Income'].iloc[0]

    print(f"{company}'s net income in 2025 was {net_income:,.0f}.")
```

```python
for company in ['Microsoft', 'Apple', 'Tesla']:
    net_income = df[
        (df['Company'] == company) &
        (df['Year'] == 2025)
    ]['Net Income'].iloc[0]

    responses[f"What was {company}'s net income in 2025?"] = (
        f"{company}'s net income in 2025 was {net_income:,.0f}."
    )
```

```python
responses #checking responses and queries
```

```python
#retrieving revenue change for all three companies.
for company in ['Microsoft', 'Apple', 'Tesla']:
    revenue_growth = df[
        (df['Company'] == company) &
        (df['Year'] == 2025)
    ]['Revenue Growth (%)'].iloc[0]

    print(
        f"{company}'s revenue changed by "
        f"{revenue_growth:.2f}% from 2024 to 2025."
    )
```

```python
#creating Query 3 and adding to responses: Revenue change from 2024 to 2025.
for company in ['Microsoft', 'Apple', 'Tesla']:
    revenue_growth = df[
        (df['Company'] == company) &
        (df['Year'] == 2025)
    ]['Revenue Growth (%)'].iloc[0]

    responses[f"How did {company}'s revenue change from 2024 to 2025?"] = (
        f"{company}'s revenue changed by {revenue_growth:.2f}% "
        f"from 2024 to 2025."
    )
```

```python
responses #checking responses and queries
```

```python
#retrieving net income change for all three companies.
for company in ['Microsoft', 'Apple', 'Tesla']:
    net_income_growth = df[
        (df['Company'] == company) &
        (df['Year'] == 2025)
    ]['Net Income Growth (%)'].iloc[0]

    print(
        f"{company}'s net income changed by "
        f"{net_income_growth:.2f}% from 2024 to 2025."
    )
```

```python
#creating Query 4 and adding to responses: Net income change.
for company in ['Microsoft', 'Apple', 'Tesla']:
    net_income_growth = df[
        (df['Company'] == company) &
        (df['Year'] == 2025)
    ]['Net Income Growth (%)'].iloc[0]

    responses[
        f"How did {company}'s net income change from 2024 to 2025?"
    ] = (
        f"{company}'s net income changed by "
        f"{net_income_growth:.2f}% from 2024 to 2025."
    )
```

```python
responses
```

```python
#Calculate the average revenue
average_revenue = df.groupby('Company')['Total Revenue'].mean()

average_revenue
```

```python
#creating Query 5 and adding to responses: Average Revenue.
for company in ['Microsoft', 'Apple', 'Tesla']:
    avg_revenue = df[
        df['Company'] == company
    ]['Total Revenue'].mean()

    responses[
        f"What was the average revenue for {company} from 2023 to 2025?"
    ] = (
        f"{company}'s average revenue from 2023 to 2025 "
        f"was {avg_revenue:,.2f}."
    )
```

```python
responses #checking all responses and queries
```

## Step 3: Building chatbot logic

```python
#creating a function for the chatbot to check queries
def simple_chatbot(user_query):
    if user_query in responses:
        return responses[user_query]
    else:
        return "Sorry, I can only provide information on predefined queries."
```

```python
# Testing a query
simple_chatbot("What was Apple's revenue in 2025?")
```

```python
# Testing an unsupported query
simple_chatbot("What is Apple's stock price?")
```
